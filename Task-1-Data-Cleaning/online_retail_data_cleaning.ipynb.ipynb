```python
import pandas as pd
import numpy as np

# Load the original dataset
df = pd.read_csv("online_retail.csv", encoding="ISO-8859-1")

print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python

# Display the first five rows
display(df.head())

```

```python
# Check column names and data types
print(df.info())
```

```python
# View basic numerical statistics
display(df.describe())
```

```python
# Check missing values
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
# Check exact duplicate rows
print("\nNumber of duplicate rows:")
print(df.duplicated().sum())
```

```python
# Check column data types
print("\nColumn data types:")
print(df.dtypes)
```

```python

# Calculate missing values and percentages
missing_summary = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (df.isnull().mean() * 100).round(2)
})

display(missing_summary[missing_summary["Missing_Count"] > 0])
```

```python

# Display a sample of exact duplicate rows
duplicates = df[df.duplicated(keep=False)]

print("Rows involved in duplicate groups:", len(duplicates))
display(duplicates.head(10))
```

```python

# Convert InvoiceDate to datetime
df["InvoiceDate"] = pd.to_datetime(
    df["InvoiceDate"],
    errors="coerce"
)

# Check the conversion
print("InvoiceDate type:", df["InvoiceDate"].dtype)
print("Dates that could not be parsed:", df["InvoiceDate"].isna().sum())
```

```python
print("\nColumn data types:")
print(df.dtypes)
```

```python

# Count negative, zero and positive quantities
print("Negative quantities:", (df["Quantity"] < 0).sum())
print("Zero quantities:", (df["Quantity"] == 0).sum())

# Count negative, zero and positive prices
print("Negative unit prices:", (df["UnitPrice"] < 0).sum())
print("Zero unit prices:", (df["UnitPrice"] == 0).sum())

# Inspect records with negative unit prices
display(df[df["UnitPrice"] < 0].head(10))

# Inspect records with negative quantities
display(df[df["Quantity"] < 0].head(10))
```

```python

# 1. Check invoice numbers beginning with C
print("Invoices beginning with C:")
print(df["InvoiceNo"].astype(str).str.startswith("C").sum())

# 2. Inspect zero-price transactions
zero_price = df[df["UnitPrice"] == 0]

print("\nZero-price rows:", len(zero_price))
display(zero_price[[
    "InvoiceNo", "StockCode", "Description",
    "Quantity", "UnitPrice", "CustomerID"
]].head(10))

# 3. Check negative quantities by invoice prefix
negative_qty = df[df["Quantity"] < 0].copy()

print("\nNegative quantities by invoice prefix:")
print(
    negative_qty["InvoiceNo"]
    .astype(str)
    .str.startswith("C")
    .value_counts()
)

# 4. Check whether date conversion produced missing dates
print("\nMissing dates after conversion:",
      df["InvoiceDate"].isna().sum())
```

```python

# 1. Inspect missing values
print("Missing values:")
print(df.isnull().sum())

# 2. Check duplicate rows
print("\nExact duplicate rows:", df.duplicated().sum())

# 3. Check text fields for leading/trailing spaces
text_cols = ["InvoiceNo", "StockCode", "Description", "Country"]

for col in text_cols:
    non_missing = df[col].dropna()
    spaces = non_missing.ne(non_missing.str.strip()).sum()
    print(f"{col}: {spaces} values with surrounding spaces")

# 4. Check customer ID values
print("\nCustomerID data type:", df["CustomerID"].dtype)
print("Non-missing CustomerIDs that are not whole numbers:",
      ((df["CustomerID"].dropna() % 1) != 0).sum())

# 5. Inspect negative quantities without C invoice prefix
negative_without_c = df[
    (df["Quantity"] < 0) &
    (~df["InvoiceNo"].astype(str).str.startswith("C"))
]

display(negative_without_c.head(10))
```

```python

# Keep the original data unchanged
df_raw = df.copy()

# Create a working copy
df_clean = df.copy()

# 1. Remove leading and trailing spaces from text columns
text_cols = ["InvoiceNo", "StockCode", "Description", "Country"]

for col in text_cols:
    df_clean[col] = df_clean[col].str.strip()

# 2. Remove exact duplicate rows
rows_before = len(df_clean)
df_clean = df_clean.drop_duplicates().copy()
duplicates_removed = rows_before - len(df_clean)

# 3. Convert CustomerID to nullable integer
df_clean["CustomerID"] = df_clean["CustomerID"].astype("Int64")

print("Original rows:", len(df_raw))
print("Rows after removing duplicates:", len(df_clean))
print("Duplicate rows removed:", duplicates_removed)
print("Missing CustomerIDs retained:", df_clean["CustomerID"].isna().sum())
print("Missing descriptions retained:", df_clean["Description"].isna().sum())
print("CustomerID data type:", df_clean["CustomerID"].dtype)
```

```python

print("Remaining exact duplicates:", df_clean.duplicated().sum())

print("\nRemaining surrounding spaces:")
for col in text_cols:
    non_missing = df_clean[col].dropna()
    print(col, ":", non_missing.ne(non_missing.str.strip()).sum())

print("\nMissing values after cleaning:")
print(df_clean.isnull().sum())

print("\nCleaned data types:")
print(df_clean.dtypes)

print("\nCleaned dataset shape:", df_clean.shape)
```

```python

# Check for remaining data-quality issues
print("Negative quantities:",
      (df_clean["Quantity"] < 0).sum())

print("Zero unit prices:",
      (df_clean["UnitPrice"] == 0).sum())

print("Negative unit prices:",
      (df_clean["UnitPrice"] < 0).sum())

print("Missing invoice dates:",
      df_clean["InvoiceDate"].isna().sum())

print("Missing descriptions:",
      df_clean["Description"].isna().sum())

print("Missing CustomerIDs:",
      df_clean["CustomerID"].isna().sum())

# Inspect the cleaned data
display(df_clean.head())
```

```python

output_file = "cleaned_online_retail.csv"

df_clean.to_csv(output_file, index=False)

print("Cleaned dataset saved successfully!")
print("File:", output_file)
```

```python

import zipfile
import os

csv_file = "cleaned_online_retail.csv"
zip_file = "cleaned_online_retail.zip"

with zipfile.ZipFile(
    zip_file,
    "w",
    compression=zipfile.ZIP_DEFLATED
) as zf:
    zf.write(csv_file, arcname="cleaned_online_retail.csv")

csv_size = os.path.getsize(csv_file) / (1024 ** 2)
zip_size = os.path.getsize(zip_file) / (1024 ** 2)

print(f"CSV size: {csv_size:.2f} MB")
print(f"ZIP size: {zip_size:.2f} MB")
```
