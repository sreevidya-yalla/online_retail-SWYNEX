## 1. Load the Raw Dataset

Load the original CSV into a Pandas DataFrame.
The raw dataset will remain unchanged.

```python
RAW_FILE ="../online_retail.csv"
df = pd.read_csv(RAW_FILE, low_memory=False)

print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
# Count missing values in each column
missing_values = df.isnull().sum()

# Display the results
print(missing_values)
```

```python
# Calculate the percentage of missing values per column
missing_percent = (df.isnull().sum() / len(df)) * 100

# Display percentages rounded to two decimal places
print(missing_percent.round(2))
```

```python
# Count rows that are exact duplicates
duplicate_count = df.duplicated().sum()

# Display the duplicate count
print("Duplicate rows:", duplicate_count)
```

```python
# Display the first few duplicate rows
df[df.duplicated()].head(10)
```

```python
# Create a separate copy for cleaning
df_clean = df.copy()

# Remove exact duplicate rows and keep the first occurrence
df_clean = df_clean.drop_duplicates()

# Display the number of rows after removing duplicates
print("Rows before cleaning:", len(df))
print("Rows after removing duplicates:", len(df_clean))
print("Duplicate rows remaining:", df_clean.duplicated().sum())
```

```python
# Count missing descriptions after duplicate removal
print("Missing descriptions:", df_clean["Description"].isnull().sum())

# Display some rows with missing descriptions
df_clean[df_clean["Description"].isnull()].head(10)
```

```python

# Find descriptions that contain only spaces
blank_descriptions = (
    df_clean["Description"].notna()
    & df_clean["Description"].str.strip().eq("")
)

# Count descriptions that are blank or contain only spaces
print("Blank descriptions:", blank_descriptions.sum())
```

```python
    # Remove extra spaces around existing descriptions
df_clean["Description"] = df_clean["Description"].str.strip()

# Check the number of missing descriptions again
print("Missing descriptions:", df_clean["Description"].isnull().sum())
```

```python
# Find descriptions containing question marks
question_mark_rows = df_clean[
    df_clean["Description"].str.contains("?", regex=False, na=False)
]

# Display the suspicious descriptions
print("Rows containing question marks:", len(question_mark_rows))
display(question_mark_rows[["StockCode", "Description"]].head(20))
```

```python
# Count the frequency of each suspicious description
print(question_mark_rows["Description"].value_counts().head(20))
```

```python
# Check other descriptions associated with the same stock codes
suspicious_codes = question_mark_rows["StockCode"].unique()

related_rows = df_clean[df_clean["StockCode"].isin(suspicious_codes)]

# Show the different descriptions found for these stock codes
print(
    related_rows.groupby("StockCode")["Description"]
    .nunique(dropna=True)
    .sort_values(ascending=False)
    .head(20)
)
```

```python
# Identify descriptions made entirely of question marks
only_question_marks = df_clean["Description"].str.fullmatch(
    r"\?+", na=False
)

# Replace those unclear descriptions with missing values
df_clean.loc[only_question_marks, "Description"] = np.nan

# Count missing descriptions after this change
print("Missing descriptions:", df_clean["Description"].isnull().sum())
```

```python
# Count descriptions that are still only question marks
remaining_question_marks = df_clean[
    df_clean["Description"].str.fullmatch(r"\?+", na=False)
]

print("Remaining question-mark-only descriptions:",
      len(remaining_question_marks))
```

```python
# Display some invoice dates
print(df_clean["InvoiceDate"].head())

# Check the current data type
print(df_clean["InvoiceDate"].dtype)
```

```python
# Convert invoice dates from text to datetime
df_clean["InvoiceDate"] = pd.to_datetime(
    df_clean["InvoiceDate"],
    errors="coerce"
)

# Display the new data type
print("InvoiceDate type:", df_clean["InvoiceDate"].dtype)
```

```python
# Count dates that could not be converted
print("Invalid or missing dates:", df_clean["InvoiceDate"].isna().sum())
```

```python
# Display a few customer IDs
print(df_clean["CustomerID"].head(10))

# Check the current data type
print("Current type:", df_clean["CustomerID"].dtype)

# Count missing customer IDs
print("Missing customer IDs:", df_clean["CustomerID"].isna().sum())
```

```python
# Select customer IDs that are present
customer_ids = df_clean["CustomerID"].dropna()

# Check whether any customer ID has a fractional part
has_fractional_ids = (customer_ids % 1 != 0).any()

print("Contains fractional customer IDs:", has_fractional_ids)
```

```python
# Convert customer IDs to integers while preserving missing values
df_clean["CustomerID"] = df_clean["CustomerID"].astype("Int64")

# Display the new data type
print("New type:", df_clean["CustomerID"].dtype)

# Display a few customer IDs
print(df_clean["CustomerID"].head(10))

# Count missing customer IDs
print("Missing customer IDs:", df_clean["CustomerID"].isna().sum())
```

```python
# Display basic statistics for quantity
print(df_clean["Quantity"].describe())

# Count transactions with negative quantities
print("Negative quantities:", (df_clean["Quantity"] < 0).sum())

# Count transactions with zero quantity
print("Zero quantities:", (df_clean["Quantity"] == 0).sum())
```

```python
# Display basic statistics for unit price
print(df_clean["UnitPrice"].describe())

# Count zero prices
print("Zero prices:", (df_clean["UnitPrice"] == 0).sum())

# Count negative prices
print("Negative prices:", (df_clean["UnitPrice"] < 0).sum())
```

```python
# Select transactions with negative quantities
negative_quantity_rows = df_clean[df_clean["Quantity"] < 0]

# Display examples of these transactions
display(
    negative_quantity_rows[
        ["InvoiceNo", "StockCode", "Description",
         "Quantity", "UnitPrice", "CustomerID"]
    ].head(10)
)
```

```python

# Select transactions where the unit price is zero
zero_price_rows = df_clean[df_clean["UnitPrice"] == 0]

# Display examples for investigation
display(
    zero_price_rows[
        ["InvoiceNo", "StockCode", "Description",
         "Quantity", "UnitPrice", "CustomerID"]
    ].head(10)
)
```

```python
# Select transactions with negative unit prices
negative_price_rows = df_clean[df_clean["UnitPrice"] < 0]

# Display all such records for investigation
display(
    negative_price_rows[
        ["InvoiceNo", "StockCode", "Description",
         "Quantity", "UnitPrice", "CustomerID"]
    ]
)
```

```python
# Identify cancellation invoices
is_cancellation = df_clean["InvoiceNo"].str.startswith("C", na=False)

# Count cancellation invoice rows
print("Cancellation invoice rows:", is_cancellation.sum())

# Count rows with negative quantities
print("Negative quantity rows:", (df_clean["Quantity"] < 0).sum())
```

```python
# Flag transactions with zero unit prices
df_clean["ZeroPriceFlag"] = df_clean["UnitPrice"] == 0

# Flag transactions with negative unit prices
df_clean["NegativePriceFlag"] = df_clean["UnitPrice"] < 0

# Flag transactions with negative quantities
df_clean["NegativeQuantityFlag"] = df_clean["Quantity"] < 0

# Display the number of records in each category
print("Zero-price rows:", df_clean["ZeroPriceFlag"].sum())
print("Negative-price rows:", df_clean["NegativePriceFlag"].sum())
print("Negative-quantity rows:", df_clean["NegativeQuantityFlag"].sum())
```

```python
# Display the transactions with the largest absolute quantities
largest_quantity_rows = df_clean.loc[
    df_clean["Quantity"].abs().nlargest(10).index,
    ["InvoiceNo", "StockCode", "Description",
     "Quantity", "UnitPrice", "CustomerID"]
]

display(largest_quantity_rows)
```

```python
# Calculate revenue for each transaction
df_clean["Revenue"] = df_clean["Quantity"] * df_clean["UnitPrice"]

# Display sample transactions with revenue
display(
    df_clean[
        ["InvoiceNo", "Quantity", "UnitPrice", "Revenue"]
    ].head(10)
)
```

```python
# Check whether revenue matches quantity multiplied by unit price
expected_revenue = df_clean["Quantity"] * df_clean["UnitPrice"]

print(
    "Revenue calculation correct:",
    df_clean["Revenue"].equals(expected_revenue)
)

# Count missing revenue values
print("Missing revenue values:", df_clean["Revenue"].isna().sum())
```

```python
# Flag transactions with absolute quantities above 1,000
df_clean["HighQuantityFlag"] = df_clean["Quantity"].abs() > 1000

# Count transactions that need review
print("High-quantity transactions:", df_clean["HighQuantityFlag"].sum())

# Display a few flagged transactions
display(
    df_clean.loc[
        df_clean["HighQuantityFlag"],
        ["InvoiceNo", "StockCode", "Description",
         "Quantity", "UnitPrice", "Revenue"]
    ].head(10)
)
```

```python
# Compare row counts before and after cleaning
print("Original rows:", len(df))
print("Rows after duplicate removal:", len(df_clean))

# Calculate the number of rows removed
print("Rows removed:", len(df) - len(df_clean))
```

```python
# Check whether any exact duplicate rows remain
print("Exact duplicate rows:", df_clean.duplicated().sum())
```

```python
# Check duplicates using the original dataset columns
original_columns = df.columns.tolist()

print(
    "Duplicates in original columns:",
    df_clean.duplicated(subset=original_columns).sum()
)
```

```python
# Count missing values in each column
print(df_clean.isnull().sum())
```

```python
# Check the data types after cleaning
print(df_clean.dtypes)
```

```python
# Confirm that revenue is calculated correctly
print(
    "Revenue calculation correct:",
    df_clean["Revenue"].equals(
        df_clean["Quantity"] * df_clean["UnitPrice"]
    )
)

# Check unusual transaction counts
print("Zero-price rows:", df_clean["ZeroPriceFlag"].sum())
print("Negative-price rows:", df_clean["NegativePriceFlag"].sum())
print("Negative-quantity rows:", df_clean["NegativeQuantityFlag"].sum())
print("High-quantity rows:", df_clean["HighQuantityFlag"].sum())
```

```python
# Save the cleaned dataset without the DataFrame index
output_file = "cleaned_online_retail.csv"

df_clean.to_csv(output_file, index=False)

print("Cleaned dataset saved successfully!")
print("File name:", output_file)
print("Rows:", len(df_clean))
print("Columns:", len(df_clean.columns))
```

```python
# Load the saved CSV to verify that it can be read
df_saved = pd.read_csv(output_file)

print("Saved file shape:", df_saved.shape)
print("Columns:", df_saved.columns.tolist())
print("Missing values in saved file:", df_saved.isnull().sum().sum())
```

```python
# Check the saved file's important columns
print("Invoice dates missing:", df_saved["InvoiceDate"].isnull().sum())
print("Customer IDs missing:", df_saved["CustomerID"].isnull().sum())
print("Descriptions missing:", df_saved["Description"].isnull().sum())

# Check whether revenue was saved correctly
expected_revenue = df_saved["Quantity"] * df_saved["UnitPrice"]

print(
    "Revenue saved correctly:",
    np.allclose(df_saved["Revenue"], expected_revenue)
)

# Confirm no duplicate records remain
print(
    "Duplicate rows:",
    df_saved.duplicated(subset=original_columns).sum()
)
```
